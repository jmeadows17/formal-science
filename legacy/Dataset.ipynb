````python
import re
from typing import List, Dict

def split_qas_markdown(text: str) -> List[Dict[str, str]]:
    """
    Finds blocks like Q6 ... A6 ... Q7 ... A7 ... (markdown friendly)
    and returns [{"n": 6, "question": "...", "answer": "..."}, ...]
    """
    # Accept optional **bold**, optional separators (: . ) -), and whitespace
    pattern = re.compile(
        r"""
        \*{0,2}\s*Q(?P<num>\d+)\s*[:.)-]?\s*\*{0,2}\s*   # Qn
        (?P<q>.*?)                                       # question
        \b\*{0,2}\s*A(?P=num)\s*[:.)-]?\s*\*{0,2}\s*     # An with same n
        (?P<a>.*?)                                       # answer
        (?=\s*\*{0,2}\s*Q\d+\s*[:.)-]?\s*\*{0,2}|$)      # next Q or end
        """,
        re.DOTALL | re.VERBOSE | re.IGNORECASE,
    )


    def _clean(s: str) -> str:
        s = s.strip()
        # Remove leading/trailing markdown dividers nearby
        s = re.sub(r"^\s*[-*_]{3,}\s*$", "", s, flags=re.MULTILINE)
        # Normalise excessive blank lines
        s = re.sub(r"\n{3,}", "\n\n", s)
        return s.strip()

    qas = []
    for m in pattern.finditer(text):
        n = int(m.group("num"))
        q = _clean(m.group("q"))
        a = _clean(m.group("a"))
        qas.append({"n": n, "question": q, "answer": a})
    return qas
````

````python
def extract_comments(example):
    comments = []
    splits = example.split('/-')
    for s in splits[1:]:
        c = '/-' + s.split('-/')[0] + '-/'
        if is_bad(c):
            comments.append(c)
    
    # must add these after as they are possibly subsets of the above
    lines = example.splitlines(keepends=True)
    for c in lines:
        if is_bad(c):
            comments.append(c)

    return comments

# exclude all comments with a count of 1 for each marker
def is_bad(comment):
    markers = ['C1','C2','C3','C4','C5']
    counts = [comment.count(m) for m in markers]
    num_markers = len(markers) - counts.count(0)
    if num_markers > 1:
        return True
    else:
        return False
    
# extract imports
def get_imports(example):
    lines = example.splitlines(keepends=True)
    for i, line in enumerate(lines):
        if 'C1' in line:
            break

    imports = ''.join(lines[:i])
    return imports


def extract_lean(example):
    
    comments = extract_comments(example)

    for c in comments:
        example = example.replace(c,'')

    imports = get_imports(example)

    lines = example.splitlines(keepends=True)
    for i2, line in enumerate(lines):
        if 'C2' in line:
            break
    for i3, line in enumerate(lines):
        if 'C3' in line:
            break
    for i4, line in enumerate(lines):
        if 'C4' in line:
            break
    for i5, line in enumerate(lines):
        if 'C5' in line:
            break

    c1 = ''.join(lines[:i2])
    c2 = imports + ''.join(lines[i2:i3])
    c3 = imports + ''.join(lines[i3:i4])
    c4 = imports + ''.join(lines[i4:i5])
    c5 = imports + ''.join(lines[i5:])

    return [c1, c2, c3, c4, c5]

def remove_unterminated_block_open_tokens(s: str) -> str:
    """
    Remove ONLY the opening token '/-' or '/-!' when that token is not matched
    by any closing '-/' later in the same string. Properly nested/closed blocks
    are preserved. (We do a single pass with a stack.)
    """
    n = len(s)
    stack = []          # list of (pos, token_len) for '/-' or '/-!' that are still open
    i = 0
    while i < n:
        if s.startswith("/-", i):
            tok_len = 3 if i + 2 < n and s[i+2] == "!" else 2
            stack.append((i, tok_len))
            i += tok_len
            continue
        if s.startswith("-/", i):
            if stack:
                stack.pop()   # close the most recent opener
            i += 2
            continue
        i += 1

    # Any entries left in stack are unterminated; remove just those opening tokens.
    if not stack:
        return s
    # Build a mask of chars to keep
    to_delete = set()
    for pos, tok_len in stack:
        to_delete.update(range(pos, pos + tok_len))
    out_chars = [ch for idx, ch in enumerate(s) if idx not in to_delete]
    return "".join(out_chars)


def clean_chunks(chunks):
    """
    Apply the unterminated-opening cleanup to each chunk returned by extract_lean.
    """
    return [remove_unterminated_block_open_tokens(c) for c in chunks]
````

````python
example = """
import Mathlib.Tactic
import Mathlib.Algebra.Algebra.Tower
import Mathlib.Analysis.Complex.Exponential
import Mathlib.Data.Complex.Basic
import Mathlib.Analysis.SpecialFunctions.Trigonometric.Deriv
import Mathlib.Topology.Basic

namespace PhysicsDerivations

/-!
## C4: Heisenberg Picture Unequal-Time Commutator
We define our own commutator `⁅x,y⁆ = x*y - y*x` to avoid name clashes.
-/

section HeisenbergPicture

variable {A : Type*} [Ring A] [Algebra ℂ A]

-- Our commutator (avoid the name `comm`).
def commutator (x y : A) : A := x * y - y * x
notation "⁅" x ", " y "⁆" => commutator x y

@[simp] lemma commutator_self (x : A) : ⁅x, x⁆ = 0 := by
  unfold commutator; abel

@[simp] lemma commutator_swap (x y : A) : ⁅y, x⁆ = -⁅x, y⁆ := by
  unfold commutator; abel

@[simp] lemma commutator_add_left (x x' y : A) :
    ⁅x + x', y⁆ = ⁅x, y⁆ + ⁅x', y⁆ := by
  unfold commutator
  simp [add_mul, mul_add, sub_eq_add_neg, add_comm, add_left_comm, add_assoc]

@[simp] lemma commutator_add_right (x y y' : A) :
    ⁅x, y + y'⁆ = ⁅x, y⁆ + ⁅x, y'⁆ := by
  unfold commutator
  simp [add_mul, mul_add, sub_eq_add_neg, add_comm, add_left_comm, add_assoc]

@[simp] lemma commutator_smul_left (a : ℂ) (x y : A) :
    ⁅a • x, y⁆ = a • ⁅x, y⁆ := by
  -- (a•x) y - y (a•x) = a•(x y) - a•(y x) = a•(x y - y x)
  unfold commutator
  calc
    (a • x) * y - y * (a • x)
        = a • (x * y) - a • (y * x) := by
          simp [Algebra.smul_mul_assoc, mul_smul]
    _   = a • (x * y - y * x) := by
          simp [smul_sub]

@[simp] lemma commutator_smul_right (a : ℂ) (x y : A) :
    ⁅x, a • y⁆ = a • ⁅x, y⁆ := by
  calc
    ⁅x, a • y⁆ = -⁅a • y, x⁆ := by simpa [commutator_swap]
    _ = -(a • ⁅y, x⁆)       := by simpa using commutator_smul_left a y x
    _ = a • (-⁅y, x⁆)       := by simpa [smul_neg]
    _ = a • ⁅x, y⁆          := by simpa [commutator_swap]

-- Heisenberg evolution with ℂ-scaling (cleaner than raw `algebraMap * x`).
noncomputable def x_heisenberg (x₀ p₀ : A) (ω m t : ℝ) : A :=
  ((Real.cos (ω * t) : ℂ) • x₀) + ((Real.sin (ω * t) / (m * ω) : ℂ) • p₀)

-- Bilinear identity: [αx + βy, γx + δy] = (αδ − βγ) • [x,y].
private lemma comm_bilinear_central (α β γ δ : ℂ) (x y : A) :
  ⁅α•x + β•y, γ•x + δ•y⁆ = (α * δ - β * γ) • ⁅x, y⁆ := by
  have h :
    ⁅α•x + β•y, γ•x + δ•y⁆ =
      α*γ • ⁅x, x⁆ + α*δ • ⁅x, y⁆ + β*γ • ⁅y, x⁆ + β*δ • ⁅y, y⁆ := by
    simp [commutator_add_left, commutator_add_right, mul_comm, mul_left_comm, mul_assoc]
  simpa [commutator_self, commutator_swap, sub_eq_add_neg, add_comm, add_left_comm, add_assoc]
    using h

/-- **C4 (smul form)**: with `⁅x₀,p₀⁆ = iℏ`,  
`⁅x(t₁), x(t₂)⁆ = (sin(ω(t₂−t₁))/(mω)) • (iℏ)` embedded in `A`. -/
theorem C4_heisenberg_commutator_smul
    (x₀ p₀ : A) (ℏ ω m : ℝ) (t₁ t₂ : ℝ)
    (h_ccr : ⁅x₀, p₀⁆ = algebraMap ℂ A (Complex.I * ℏ)) :
    ⁅x_heisenberg x₀ p₀ ω m t₁, x_heisenberg x₀ p₀ ω m t₂⁆
      = (Real.sin (ω * (t₂ - t₁)) / (m * ω) : ℂ) • algebraMap ℂ A (Complex.I * ℏ) := by
  -- Expand via bilinearity:
  have base :
    ⁅x_heisenberg x₀ p₀ ω m t₁, x_heisenberg x₀ p₀ ω m t₂⁆ =
      ((Real.cos (ω * t₁) : ℂ) * (Real.sin (ω * t₂) / (m * ω) : ℂ)
       - (Real.sin (ω * t₁) / (m * ω) : ℂ) * (Real.cos (ω * t₂) : ℂ)) • ⁅x₀, p₀⁆ := by
    simpa [x_heisenberg] using
      comm_bilinear_central
        (α := (Real.cos (ω * t₁) : ℂ))
        (β := (Real.sin (ω * t₁) / (m * ω) : ℂ))
        (γ := (Real.cos (ω * t₂) : ℂ))
        (δ := (Real.sin (ω * t₂) / (m * ω) : ℂ))
        x₀ p₀
  -- Substitute the CCR:
  have base' := by simpa [h_ccr] using base
  -- Trig: cos t₁ sin t₂ − sin t₁ cos t₂ = sin (t₂ − t₁), then divide by (m ω)
  have hR :
    (Real.cos (ω * t₁) * (Real.sin (ω * t₂) / (m * ω))
     - (Real.sin (ω * t₁) / (m * ω)) * Real.cos (ω * t₂))
      = Real.sin (ω * (t₂ - t₁)) / (m * ω) := by
    have hsin :
      Real.sin (ω * (t₂ - t₁))
        = Real.sin (ω * t₂) * Real.cos (ω * t₁)
          - Real.cos (ω * t₂) * Real.sin (ω * t₁) := by
      simpa [sub_eq_add_neg] using Real.sin_sub (ω * t₂) (ω * t₁)
    have :
      Real.cos (ω * t₁) * Real.sin (ω * t₂)
        - Real.sin (ω * t₁) * Real.cos (ω * t₂)
        = Real.sin (ω * (t₂ - t₁)) := by
      simpa [mul_comm, mul_left_comm, mul_assoc, sub_eq_add_neg] using hsin.symm
    simpa [div_eq_mul_inv, mul_comm, mul_left_comm, mul_assoc] using
      congrArg (fun z => (1 / (m * ω)) * z) this
  -- Coerce to ℂ:
  have hC :
    ((Real.cos (ω * t₁) : ℂ) * (Real.sin (ω * t₂) / (m * ω) : ℂ)
     - (Real.sin (ω * t₁) / (m * ω) : ℂ) * (Real.cos (ω * t₂) : ℂ))
      = (Real.sin (ω * (t₂ - t₁)) / (m * ω) : ℂ) := by
    simpa using congrArg Complex.ofReal hR
  -- Done:
  simpa [hC] using base'

/-- **C4 (product-of-`algebraMap` form)** to match your target style. -/
theorem C4_heisenberg_commutator
    (x₀ p₀ : A) (ℏ ω m : ℝ) (t₁ t₂ : ℝ)
    (h_ccr : ⁅x₀, p₀⁆ = algebraMap ℂ A (Complex.I * ℏ)) :
    ⁅x_heisenberg x₀ p₀ ω m t₁, x_heisenberg x₀ p₀ ω m t₂⁆
      = algebraMap ℂ A (Complex.I * ℏ / (m * ω))
        * algebraMap ℂ A (Real.sin (ω * (t₂ - t₁))) := by
  -- Start from the smul form and rewrite smul via `algebraMap`.
  have h := C4_heisenberg_commutator_smul x₀ p₀ ℏ ω m t₁ t₂ h_ccr
  simpa [Algebra.smul_def, map_mul, div_eq_mul_inv,
         mul_comm, mul_left_comm, mul_assoc] using
    (by simpa [Algebra.smul_def, map_mul, div_eq_mul_inv,
               mul_comm, mul_left_comm, mul_assoc] using h)

end HeisenbergPicture
end PhysicsDerivations


"""
````
